```python
query = """
SELECT
    CONCAT(c.FirstName, ' ', c.LastName) AS FullName,
    c.Email,
    c.Phone,
    o.TotalCost
FROM customers AS c
INNER JOIN orders AS o
    ON c.CustomerID = o.CustomerID
WHERE o.TotalCost > 60;
"""

cursor.execute(query)

results = cursor.fetchall()

print(results)
```
